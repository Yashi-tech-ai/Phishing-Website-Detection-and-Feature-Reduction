```python
import pandas as pd
from scipy.io import arff

data, meta = arff.loadarff("../data/Training Dataset.arff")
df = pd.DataFrame(data)

for column in df.columns:
    if df[column].dtype == "object":
        df[column] = df[column].str.decode("utf-8").astype(int)

df.shape
```

```python
X = df.drop(columns="Result")
y = df["Result"]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
print("Feature data types:")
print(X.dtypes.value_counts())

print("\nMissing values:")
print(X.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining class proportions:")
print(y_train.value_counts(normalize=True))

print("\nTesting class proportions:")
print(y_test.value_counts(normalize=True))
```

### Preprocessing Summary

The dataset contains 11,055 instances and 30 numerical input features with no missing values. The target variable `Result` contains two classes.

The data was divided into training and testing sets using an 80:20 stratified split. This resulted in 8,844 training samples and 2,211 testing samples. Stratification preserved the class distribution across both sets, with approximately 55.7% belonging to class 1 and 44.3% belonging to class -1.

No missing-value imputation or categorical encoding was required because the dataset contains complete numerical features. Feature scaling will be handled separately for algorithms that require it, using training data only to avoid data leakage.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

```python
print("Mean of first 5 scaled features:")
print(X_train_scaled[:, :5].mean(axis=0))

print("\nStandard deviation of first 5 scaled features:")
print(X_train_scaled[:, :5].std(axis=0))
```
